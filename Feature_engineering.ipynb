```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import numpy as np 
```

```python
dataset = pd.read_csv("AmesHousing.csv")
```

# Feature Engineering 

### step1 : Only Conceptual (not coding)

```python
dataset['House Age'] = dataset['Yr Sold'] - dataset['Year Built']   # House Age is better than Year Built because it shows the how much the house is old 
```

```python
dataset['Year since Remodel'] = dataset['Yr Sold'] - dataset['Year Remod/Add']
```

```python
dataset['Is_Remodeled'] = (dataset['Year Built'] != dataset['Year Remod/Add']).astype(int)
```

```python
dataset['Total Living Area'] = dataset['Gr Liv Area']+dataset['Total Bsmt SF']
```

```python
dataset['Total Finished Area'] = dataset['Gr Liv Area'] + dataset['BsmtFin SF 1'] + dataset['BsmtFin SF 2']
```

```python
dataset['Has_Basement'] = dataset['Total Bsmt SF'].apply(lambda x: 1 if x > 0 else 0)
```

```python
dataset['Total Bathrooms'] = dataset['Bsmt Full Bath']+0.5*dataset['Bsmt Half Bath']+dataset['Full Bath']+0.5*dataset['Half Bath']
```

```python
dataset['Garage Age'] = dataset['Yr Sold'] - dataset['Garage Yr Blt']
```

We Check the Garage Age only if Garage exists 

```python
dataset['Garage Exists'] = dataset['Garage Area'].apply(lambda x: 1 if x>0 else 0)
```

```python
dataset['Quality_size_score'] = dataset['Overall Qual']*dataset['Gr Liv Area']
```

```python
dataset['Non_Bedroom_Rooms'] = dataset['TotRms AbvGrd'] - dataset['Bedroom AbvGr']
```
